## ⏪ 복습: 지난 시간까지

- **그래프 데이터 모델(LPG)**: 개체는 **노드**, 개체 사이의 연결은 **관계**, 딸린 값은 **속성**. 레이블로 노드의 종류를 구분했습니다.
- **그래프 순회**: 이웃을 따라가며(1홉·2홉) 답을 찾았습니다. RDB 라면 여러 번 조인해야 할 질문을 연결을 따라 바로 풀 수 있었죠.
- 그때는 dict·list 로 **직접** 그래프를 만들었지만, 규모가 커지면 전용 DB 가 필요합니다. 그게 Neo4j 입니다.

- [ ] 예제 그래프 **Movies** 를 불러온다(제공 스크립트 또는 내장 안내).
- [ ] 제공된 조회를 실행해 **그래프가 답하는 첫 경험**을 하고, 결과를 파이썬으로 읽는다.

# 1. Neo4j 설치하기

## Neo4j 란?
**Neo4j** 는 그래프를 저장·조회하는 데 특화된 **그래프 데이터베이스**입니다. 표(행·열) 대신 **노드와 관계**를 저장 구조의 기본 단위로 삼아, "누가 무엇과 어떻게 연결되어 있나"를 빠르게 답합니다. 질문은 **Cypher** 라는 쿼리 언어로 던집니다(다음 단원에서 학습).

설치 방법은 두 가지입니다. **처음이라면 Neo4j Desktop 을 권장**합니다(내 컴퓨터에 깔아 오프라인으로 실습).

## 방법 A: Neo4j Desktop (권장, 로컬 설치)
Desktop 에서 만드는 데이터베이스 한 벌을 **인스턴스(instance)** 라고 부릅니다(한 인스턴스 안에 여러 데이터베이스가 들어갈 수 있어서 이렇게 부릅니다).

1. <https://neo4j.com/deployment-center/> 에서 **Neo4j Desktop** 을 내려받아 설치합니다(Windows·macOS·Linux).
2. 첫 실행이면 화면 가운데의 **Create instance** 버튼을 누릅니다(이미 인스턴스가 있으면 이 버튼은 오른쪽 위에 있습니다).
3. 한 화면에서 아래를 채우고 **Create** 를 누릅니다.
   - **이름**: 예) `practice`
   - **버전**: 기본값(가장 최신) 그대로 두면 됩니다. 다른 버전을 고르면 내려받기가 필요해 인터넷에 연결돼 있어야 합니다.
   - **사용자·비밀번호**: 사용자는 보통 `neo4j`, 비밀번호는 원하는 값(예: `test1234`). **꼭 기억해 두세요.** 뒤에서 `.env` 에 적습니다.
4. 만들면 인스턴스가 **곧바로 시작됩니다.** 카드가 **Stopped**(멈춤)로 보일 때만 **재생 버튼**을 눌러 켜세요. 켜져 있어야 연결됩니다.
5. 접속 주소(Connection URI)는 로컬 기본이 `bolt://localhost:7687`, 사용자 이름은 `neo4j` 입니다. 인스턴스 카드에서 확인할 수 있습니다(이 값과 방금 정한 비밀번호를 뒤에서 `.env` 에 넣습니다).

> 인스턴스는 **켜 둔 동안에만** 연결됩니다. 그리고 Desktop 앱을 종료하면 함께 멈춥니다. 실습 전 항상 켜져 있는지 확인하세요.

# 2. 그래프를 눈으로 보는 도구: Query

파이썬으로 연결하기 전에, 그래프를 **눈으로** 한 번 보면 이해가 빠릅니다. Neo4j 에는 쿼리를 입력하면 **노드와 관계를 그림으로** 그려 주는 도구가 함께 들어 있습니다. Desktop·Aura 안에서는 이 도구를 **Query** 라고 부릅니다.

- **Neo4j Desktop**: 왼쪽 사이드바의 **Tools → Query**, 또는 인스턴스 카드의 **Connect 드롭다운**에서 Query 를 고릅니다(인스턴스가 켜져 있어야 합니다).
- **Aura**: 콘솔에서 인스턴스의 **Query** 를 누릅니다.

## 화면 구성
- **Cypher 편집기(가운데 위)**: 쿼리나 `:` 로 시작하는 명령을 치고 **실행(재생) 버튼**으로 실행합니다.
- **결과 프레임(아래)**: 결과를 **Graph**(노드·관계 그림)·**Table**(표)·**RAW** 로 바꿔 볼 수 있습니다. 결과 프레임 안에서 쿼리를 고쳐 다시 실행할 수도 있습니다.
- **사이드바(왼쪽)**: 아이콘으로 패널을 바꿔 엽니다. 패널은 다섯 개이고 맨 아래 톱니는 설정입니다.
  - **Database information**: 지금 데이터베이스의 노드·관계 수, 레이블·관계 종류·속성 키 목록(어떤 데이터가 들어 있는지 한눈에). 오늘 실제로 여는 패널은 이 하나입니다.
  - 나머지 넷은 **Saved Cypher**(저장한 쿼리)·**History**(실행 기록)·**Cypher Reference**(문법 빠른 참조)·**Parameters**(쿼리에 넘길 파라미터)입니다.

# 3. 파이썬에서 첫 연결

이제 파이썬에서 데이터베이스에 연결합니다. 아래 **준비 셀**은 `.env` 파일의 접속 정보(`NEO4J_URI`·`NEO4J_USER`·`NEO4J_PASSWORD`)로 연결하고, 쿼리를 실행하는 도우미 `run_cypher(...)` 를 만들어 줍니다.

**연결 정보는 `.env` 로만 설정합니다**. 노트북에 주소·비밀번호를 직접 적지 마세요. `.env.example` 을 복사해 `.env` 로 만들고, 1절에서 정한 값(로컬은 `bolt://localhost:7687`)을 채우면 됩니다.

````python
# Neo4j 연결: 이 셀은 실행만 하세요(내용은 이해하지 않아도 됩니다).
# - .env 의 NEO4J_URI/NEO4J_USER/NEO4J_PASSWORD 로 데이터베이스에 연결합니다.
# - run_cypher("쿼리", 파라미터=값) 가 결과를 dict 리스트로 돌려줍니다. 이 헬퍼로 Cypher 를 실행합니다.
# - 이 단원은 그래프를 조회만 합니다(그래프를 바꾸지 않습니다).
import os

from dotenv import load_dotenv
from neo4j import GraphDatabase   # 파이썬용 공식 드라이버. 이 클래스로 접속 통로를 연다

# 1) 접속 정보 읽기: .env 에 적힌 값을 환경변수로 올린다(파일이 없으면 조용히 넘어간다)
load_dotenv(".env")       # 같은 폴더의 .env
load_dotenv("../.env")    # 정답 폴더에서 실행하는 경우

# 두 번째 인자는 .env 에 그 키가 없을 때 쓰는 기본값이다(로컬 Desktop 의 표준 주소·사용자).
# .env 를 못 읽어도 에러가 아니라 이 값으로 조용히 넘어가니, 이 셀 마지막 줄에 찍히는
# 주소가 실습 전용 DB 가 맞는지 눈으로 꼭 확인한다
NEO4J_URI = os.getenv("NEO4J_URI", "bolt://localhost:7687")
NEO4J_USER = os.getenv("NEO4J_USER", "neo4j")
NEO4J_PASSWORD = os.getenv("NEO4J_PASSWORD", "neo4j")
# 2) 드라이버 만들기: 접속 통로 하나를 노트북 전체가 나눠 쓴다(쿼리마다 새로 만들지 않는다)
driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USER, NEO4J_PASSWORD))
driver.verify_connectivity()   # 실제로 붙어 본다. 인스턴스가 꺼져 있거나 비밀번호가 틀리면 여기서 에러가 난다


# 3) 수업 내내 쓰는 헬퍼: 쿼리를 보내고 결과를 파이썬 자료형으로 바꿔 준다
def run_cypher(query, **params):
    """Cypher 실행 -> 결과를 dict 리스트로 반환(수업 공용 헬퍼)."""
    with driver.session() as session:
        # 세션은 with 블록을 벗어나면 자동으로 닫힌다. record.data() 가 결과 한 행을 dict 로 바꾼다
        return [record.data() for record in session.run(query, **params)]


print("Neo4j 연결:", NEO4J_URI)   # 이 줄이 찍히면 연결까지 성공한 것이다
````

````python
# 연결 진단: 실행만 하세요(그래프를 바꾸지 않습니다).
# 연결이 이상할 때 "어디에 무엇으로 붙었는지" 세 가지를 차례로 확인한다: 주소·서버·데이터베이스 이름
print('접속 주소:', NEO4J_URI, '/ 사용자:', NEO4J_USER)   # .env 를 못 읽었으면 기본값이 그대로 찍히니 의도한 인스턴스인지 먼저 본다
````

````python
# CALL: 서버에 내장된 프로시저를 부르는 방법. dbms.components() 는 서버가 자기 정보를 알려 준다
# versions 는 값이 여럿 담긴 리스트라 [0] 으로 첫 값만 꺼냈다
for row in run_cypher("CALL dbms.components() YIELD name, versions, edition "
                      "RETURN name, versions[0] AS version, edition"):
    print('서버:', row['name'], row['version'], row['edition'])
````

**Aura 를 만들지 않았다면 이 절은 건너뛰세요.** `AURA_*` 가 비어 있으면 아래 연결 셀에서 에러가 납니다. 로컬만으로도 뒤 단원과 과제를 전부 할 수 있습니다.

````python
# .env 에 적어 둔 원격(Aura) 접속 정보를 읽어 옵니다(실행만 하세요).
# 주소 앞머리가 연결 방식을 말해 준다. 로컬은 bolt://, Aura 는 neo4j+s:// 다
# Aura 를 안 만들었다면 원격 줄이 비어 있게 찍힌다. 그게 정상이다
AURA_URI = os.getenv("AURA_URI", "")
AURA_USER = os.getenv("AURA_USER", "neo4j")
AURA_PASSWORD = os.getenv("AURA_PASSWORD", "")
print("로컬:", NEO4J_URI)
print("원격:", AURA_URI)
````

````python
# 원격(Aura)에 붙어 조회까지 해 봅니다(실행만 하세요).
# 드라이버를 만드는 코드는 위 준비 셀과 글자까지 똑같다. 바뀌는 것은 주소 하나뿐이다
aura_driver = GraphDatabase.driver(AURA_URI, auth=(AURA_USER, AURA_PASSWORD))
aura_driver.verify_connectivity()   # 여기서 에러가 없으면 원격 연결까지 성공한 것이다
with aura_driver.session() as session:   # run_cypher 안에서 하던 것과 같은 모양이다
    rows = [record.data() for record in session.run("MATCH (n) RETURN count(n) AS cnt")]
    info = [record.data() for record in session.run("CALL dbms.components() YIELD edition RETURN edition")]
print("Aura 노드 수:", rows[0]["cnt"], "/ 에디션:", info[0]["edition"])
````

> 로컬은 `bolt://`, Aura 는 `neo4j+s://` 입니다. **`+s` 가 암호화된 연결**이라는 뜻이고, 인터넷을 건너가니 필요한 것입니다. 그런데 붙는 코드도, 조회하는 코드도 양쪽이 **똑같습니다.** **주소만 바뀝니다.**

# 5. 첫 조회: 그래프에게 물어보기

데이터가 들어왔으니 이제 **질문**을 던질 차례입니다. 아래 조회들은 모두 제공됩니다(쿼리 작성은 다음 단원). 여러분이 할 일은 **실행하고, 돌아온 결과를 파이썬으로 읽는 것**입니다.

`run_cypher("...")` 는 결과를 **dict 의 리스트**로 돌려줍니다. 지난 단원에서 다룬 바로 그 자료구조라 익숙할 겁니다. 먼저 영화가 몇 편 들어 있는지 물어봅니다.

````python
# 영화(Movie) 노드 몇 개가 들어 있는지 물어봅니다(실행만 하세요).
# MATCH (m:Movie): 레이블이 Movie 인 노드를 전부 고른다
# RETURN count(m) AS movie_count: 그 개수를 movie_count 라는 이름으로 돌려준다(AS 로 붙인 이름이 결과 dict 의 키가 된다)
result = run_cypher("MATCH (m:Movie) RETURN count(m) AS movie_count")
print(result)   # 집계는 한 행뿐이라 dict 하나짜리 리스트로 돌아온다
````

````python
# 영화 제목과 개봉연도를 몇 편 가져옵니다(실행만 하세요).
# 앞 셀과 달리 개수가 아니라 값을 꺼낸다: RETURN 에 title·released 두 속성을 적었다
# ORDER BY m.title 로 제목 알파벳순 정렬, LIMIT 5 로 앞 5편만 받는다(38편을 다 받으면 화면이 길어진다)
movies = run_cypher("MATCH (m:Movie) RETURN m.title AS title, m.released AS released "
                    "ORDER BY m.title LIMIT 5")
for row in movies:
    print(row)
````

각 `row` 는 `{'title': ..., 'released': ...}` 형태의 dict 입니다. 이제 이 결과를 **파이썬으로 후처리**해 봅니다. 그래프에서 값을 가져오는 것까지가 제공, 그 값으로 답을 만드는 건 여러분 몫입니다.

## 이번 강의 정리

| 개념 | 핵심 |
|---|---|
| Neo4j | 노드·관계를 기본 단위로 저장·조회하는 데이터베이스. 질문은 Cypher 로 |
| 설치 | Neo4j Desktop(로컬, 권장) 또는 Aura(클라우드). 인스턴스는 **켜 둬야** 연결됨 |
| 환경 | 인스턴스는 만들면 바로 켜지고, 한 번에 하나만 실행. 막히면 **진단 셀 + 에러 표** |
| 도구 | Desktop·Aura 안의 **Query** 로 그래프를 눈으로 확인 |
| 연결 | `.env` 의 접속 정보로 연결. 주소·비밀번호는 노트북에 적지 않음 |
| 로컬 대 원격 | 붙는 코드는 같고 **주소만** 다름(`bolt://` 대 `neo4j+s://`). 다음 단원부터는 로컬만 |

| 조회 결과 | `run_cypher(...)` → **dict 의 리스트**. 리스트·dict 인덱싱으로 값 추출 |

- 이 단원은 **읽기 전용**: 조회만 하고 그래프를 바꾸지 않았습니다.
- 그래프가 돌려준 결과는 이미 아는 **dict·list** 라서, 파이썬으로 얼마든지 후처리할 수 있습니다.
